```python
sns.set_theme(style="whitegrid")
```

```python
country_counts = final_pd['country'].value_counts().head(10)

plt.figure(figsize=(10, 5))

sns.barplot(
    x=country_counts.values,
    y=country_counts.index
)

plt.title("Top 10 Countries by Number of Restaurants")
plt.xlabel("Number of Restaurants")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

```python
print("Top country:", country_counts.index[0])
print("Number of restaurants:", country_counts.iloc[0])
```

```python
city_counts = final_pd['city'].value_counts().head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=city_counts.values,
    y=city_counts.index
)

plt.title("Top 10 Cities by Number of Restaurants")
plt.xlabel("Number of Restaurants")
plt.ylabel("City")
plt.tight_layout()
plt.show()
```

```python
print("City with the highest restaurants:", city_counts.index[0])
print("Restaurant count:", city_counts.iloc[0])
```

```python
rating_counts = final_pd['rating_text'].value_counts()

plt.figure(figsize=(9, 5))

sns.barplot(
    x=rating_counts.index,
    y=rating_counts.values
)

plt.title("Distribution of Restaurant Ratings")
plt.xlabel("Rating Category")
plt.ylabel("Number of Restaurants")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
print("Most common rating category:", rating_counts.index[0])
print("Number of restaurants:", rating_counts.iloc[0])
```

```python
delivery_counts = final_pd['has_online_delivery'].value_counts()

plt.figure(figsize=(6, 6))

plt.pie(
    delivery_counts.values,
    labels=delivery_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Online Delivery Availability")
plt.show()
```

```python
print(delivery_counts)
```

```python
rated_data = final_pd[
    final_pd['aggregate_rating'] > 0
]

country_rating = (
    rated_data.groupby('country')
    .agg(
        average_rating=('aggregate_rating', 'mean'),
        restaurant_count=('restaurant_id', 'count')
    )
)

country_rating = country_rating[
    country_rating['restaurant_count'] >= 20
].sort_values(
    'average_rating', ascending=False
).head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    data=country_rating.reset_index(),
    x='average_rating',
    y='country'
)

plt.title("Top 10 Countries by Average Restaurant Rating")
plt.xlabel("Average Rating")
plt.ylabel("Country")
plt.xlim(0, 5)
plt.tight_layout()
plt.show()
```
